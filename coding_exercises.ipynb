1. Simplicity

The principle of simplicity means writing code that is clear, minimal, and easy to understand.
The code has a single responsibility, it only prints text to the console. It has minimal dependencies because it requires no imports or extra logic. Its readability is high, as anyone familiar with Python can understand it instantly. This matters because simple code reduces cognitive load and makes maintenance easier.

```python

import html
import sys

def safe_print(message: str) -> None:
    """
    Prints a sanitized message to prevent injection attacks.
    """
    if not isinstance(message, str):
        raise TypeError("Message must be a string.")
    
    # Limit message length to avoid abuse
    if len(message) > 100:
        raise ValueError("Message too long.")
    
    # Escape HTML special characters
    sanitized = html.escape(message)
    print(sanitized)

if __name__ == "__main__":
    # Default message if no argument provided
    user_input = "Hello, World!"
    safe_print(user_input)
```

2. Security

The code performs input validation by checking both the type and the length of the input. It also sanitises the input using HTML.escape() to neutralise any HTML or script tags. It applies fail-safe defaults by printing “Hello, World!” when no input is provided. This is important because it helps prevent injection attacks and resource abuse.

```python

import locale

def hello_world():
    # Detect system locale
    lang = locale.getdefaultlocale()[0] or "en"
    
    greetings = {
        "en": "Hello, World!",
        "es": "¡Hola, Mundo!",
        "fr": "Bonjour, le monde!",
        "ja": "こんにちは、世界！"
    }
    
    # Fallback to English if language not supported
    message = greetings.get(lang[:2], greetings["en"])
    
    print(message)

if __name__ == "__main__":
    hello_world()
```

3. Inclusion

The code performs language detection by using the system locale to personalise output. It includes a fallback mechanism that defaults to English if the detected locale isn’t supported. This approach matters because it makes the software accessible to a global audience.

```python
def group_words_by_frequency(sentence):
    """
    Groups words in a sentence and sorts them by frequency (highest first).
    :param sentence: A string containing the sentence.
    :return: A list of tuples (word, count) sorted by count descending.
    """
    # Step 1: Normalize the sentence (lowercase)
    sentence = sentence.lower()
    
    # Step 2: Split into words
    words = sentence.split()
    
    # Step 3: Count occurrences
    word_count = {}
    for word in words:
        word_count[word] = word_count.get(word, 0) + 1
    
    # Step 4: Sort by frequency (descending)
    sorted_by_freq = sorted(word_count.items(), key=lambda x: x[1], reverse=True)
    
    return sorted_by_freq

# Example usage
sentence = "How do I love thee? Let me count the ways. I love thee to the depth and breadth and height My soul can reach, when feeling out of sight For the ends of being and ideal grace. I love thee to the level of every day’s Most quiet need, by sun and candle-light. I love thee freely, as men strive for right. I love thee purely, as they turn from praise. I love thee with the passion put to use In my old griefs, and with my childhood’s faith. I love thee with a love I seemed to lose With my lost saints. I love thee with the breath, Smiles, tears, of all my life; and, if God choose, I shall but love thee better after death."
result = group_words_by_frequency(sentence)
print(result)
```

### Understanding a Program: Parsing and Reading

```python
import webbrowser
import requests

print("Shall we hunt down an old website?")
site = input("Type a website URL: ")
era = input("Type year, month, and date, e.g., 20150613: ")
url = "http://archive.org/wayback/available?url=%s&timestamp=%s" 
(site, era)
response = requests.get(url)
data = response.json()

try:
   old_site = data["archived_snapshots"]["closest"]["url"]
   print("Found this copy: ", old_site)
   print("It should appear in your browser.")
   webbrowser.open(old_site)

except:
 print("Sorry, could not find the site.")
```

1. How many libraries were imported?
Two libraries:
- webbrowser
- requests


2. How many variables were used?
Six variables are assigned values in the code:
- site — the website URL input by the user
- era — the date (YYYYMMDD) input by the user
- url — the API request URL constructed using site and era
- response — the HTTP response from the Wayback API
- data — the parsed JSON from the response
- old_site — the URL of the closest archived snapshot (inside try)

3. How many strings appear in the program?
10 string literals appear:
- "Shall we hunt down an old website?"
- "Type a website URL: "
- "Type year, month, and date, e.g., 20150613: "
- "http://archive.org/wayback/available?url=%s&amp;timestamp=%s"
- "archived_snapshots"
- "closest"
- "url"
- "Found this copy: "
- "It should appear in your browser."
- "Sorry, could not find the site."

4. How many times is something printed? What was printed?
There are 4 print() statements in the code.
- print("Shall we hunt down an old website?")
- print("Found this copy: ", old_site)  → prints the message and the snapshot URL

### Text Processing Example: Word Frequency Counting

```python
from collections import Counter
import re

def process_text(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        text = f.read().lower()

    # Remove punctuation and split into words
    words = re.findall(r'\w+', text)
    word_counts = Counter(words)

    for word, count in word_counts.most_common(10):
        print(f"{word}: {count}")

process_text("some_text.txt")
```

This example shows how Python can process text by counting how often each word appears in a document. The code reads a text file, cleans the text by converting it to lowercase, removes punctuation, splits the text into words, and then counts the frequency of each word. Finally, it prints the ten most common words.

I can recognise several Python concepts from earlier exercises. The program uses imports to add additional tools. It uses a function to organise the code into reusable steps. Inside the function, there are several variables, such as text, words, and word_counts. The code also uses a for loop to print the most common words, and several print statements to show the output. The file path "some_text.txt" is an example of a string.

Overall, this example helps me see how simple Python concepts, like variables, loops, and functions, can be used to perform basic natural language processing tasks.
