#### 1. Variables

```python
# Store a message in a variable
greeting = "Hello, world!"

# Print the message using the variable
print(greeting)
```

**Step-By-Step**

greeting = "Hello, world!"
  - The coding above is a variable named 'greeting'.
  - A variable is like a labelled box where you can store information.
  - In this example, the information is the text "Hello, world!". In Python it is called 'string'.

print(greeting)
  - Instead of printing the text directly, it print the value stored in the variable.
  - This shows that variables can hold data and be reused anywhere in the program.

**Key Aspects**
 - It makes it able to chnage the message by updating the variable without modifying the 'print ()' satement, which makess the code flexible.
 - Text data must be enclosed in quotes (" " or ' ')

#### 2. Functions

```python
# Define a function that prints a greeting
def say_hello():
    print("Hello, world!")

# Call the function to run it
say_hello()
```

**Step-By-Step**

def say_hello():
 - In Python 'def' is a special keyword that means define a function.
 - 'say_hello' is the name of the function. You can choose any name, but it should describe what the function does.
 - '()' means this function does not take any input yet.
 - ':' tells Python that the next indented lines belong to this function.

print("Hello, world!")
 - This line is indented (spaces at the start) because it is inside the function.
 - It tells Python what the function should do → print "Hello, world!".

say_hello()
 - Writing the function name followed by () runs the function.
 - If you do not call the function, nothing happens because defining a function only stores it for later use.



**Key Aspects**

Functions in Python are essential because they allow you to organise code into reusable blocks, making programs easier to read, maintain, and debug. A function is defined using the 'def' keyword followed by a name and parentheses, and its body must be indented to show which statements belong to it. Defining a function does not execute it, you must call the function by writing its name followed by parentheses. This separation between definition and execution is important because it lets you reuse the same logic multiple times without rewriting the code. Moreover, functions make programs more flexible, as they can later accept inputs and return outputs, enabling dynamic behaviour.




#### 3. Other Initial prompts to explore

##### **Initial prompt to understand loops**

```python
# Print "Hello, world!" 5 times using a loop
for i in range(4):
    print("Hello, world!")
```

**Step-By-Step**

for i in range(4):
 - This is the start for the loop in Python.
 - 'i' is a loop variable that changes each time the loop runs.
 - 'range(4)' creates a sequence of numbers: 0, 1, 2, 3, 4.
 - The loop will run 4 times, once for each number in that sequence.
 - ':' tells Python that the next indented lines belong to the loop.

print("Hello, world!")
 - This line is indented, so it runs every time the loop repeats.
 - Since the loop runs 4 times, "Hello, world!" will be printed 4 times.


**Key Aspects**

Loops repeat code automatically without writing it multiple times. Everything inside the loop must be indented. Moreover, 'range(#)' controls how many times the loop runs. The loop variable (i) can be used if you want to show the count.
