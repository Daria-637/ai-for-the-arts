#### 2. Functions

```python
# Define a function that prints a greeting
def say_hello():
    print("Hello, world!")

# Call the function to run it
say_hello()
```

**Step-By-Step**

def say_hello():
 - In Python 'def' is a special keyword that means define a function.
 - 'say_hello' is the name of the function. You can choose any name, but it should describe what the function does.
 - '()' means this function does not take any input yet.
 - ':' tells Python that the next indented lines belong to this function.

print("Hello, world!")
 - This line is indented (spaces at the start) because it is inside the function.
 - It tells Python what the function should do → print "Hello, world!".

say_hello()
 - Writing the function name followed by () runs the function.
 - If you do not call the function, nothing happens because defining a function only stores it for later use.



**Key Aspects**

Functions in Python are essential because they allow you to organise code into reusable blocks, making programs easier to read, maintain, and debug. A function is defined using the 'def' keyword followed by a name and parentheses, and its body must be indented to show which statements belong to it. Defining a function does not execute it, you must call the function by writing its name followed by parentheses. This separation between definition and execution is important because it lets you reuse the same logic multiple times without rewriting the code. Moreover, functions make programs more flexible, as they can later accept inputs and return outputs, enabling dynamic behaviour.

##### **Initial prompt to understand data structures such as dictionaries (3)**

```python
# Step 1: Create a dictionary of greetings in different languages
greetings = {
    "English": "Hello, world! ʚ❤︎ɞ",
    "Spanish": "¡Hola, mundo! ʚ❤︎ɞ",
    "French": "Bonjour, le monde! ʚ❤︎ɞ",
    "Japanese": "こんにちは, 世界! ʚ❤︎ɞ"
}

# Step 2: Loop through the dictionary and print each greeting
for language, message in greetings.items():
    print(f"{language}: {message}")
```

greetings = {
    "English": "Hello, world! ʚ❤︎ɞ",

 - A dictionary stores data in key-value pairs. Keys (like "English", "Japanese") act like labels. Values (like "Hello, world! ʚ❤︎ɞ", "こんにちは, 世界! ʚ❤︎ɞ") are the data associated with those keys.
 - Dictionaries use curly braces '{ }' and separate keys and values with a colon ':'.


for language, message in greetings.items():
    print(f"{language}: {message}")
 - '.items()' gives both the key and value for each pair in the dictionary.
 - language stores the key (e.g., "English"), and message stores the value (e.g., "Hello, world! ʚ❤︎ɞ").
 - 'print(f"{language}: {message}")' uses an f-string to format the output nicely.

**Key Aspects**

Dictionaries store data as key-value pairs, allowing you to easily look up values by their keys. You can loop through keys and values using '.items()'. They are useful for organising data with clear labels and corresponding values.

##### **Initial prompt to understand how to use conditional if, else, elif statements (6)**

```python
# Step 1: Ask the user for their favorite number
number = int(input("Enter your favorite number: "))

# Step 2: Use if, elif, and else to respond
if number > 10:
    print("Hello, world! That's a big number!")
elif number == 10:
    print("Hello, world! That's exactly ten!")
else:
    print("Hello, world! That's a small number!")
```

**Step-By-Step**

number = int(input("Enter your favorite number: "))
 - 'input()' asks the user to type something.
 - 'int()' converts the input from text to an integer.
 - The number is stored in the variable number.


if number > 10:
    print("Hello, world! That's a big number!")
 - If the first condition was false, this checks if the number equals 10.


else:
    print("Hello, world! That's a small number!")
 - 'else' runs if none of the above are true


**Key Aspects**

'if', 'elif', 'else' let you make decisions in your code. Conditions use comparison operators like '>', '<', '=='. Moreover, Indentation matters. All code inside each block must be indented. Only one block runs based on the first true condition.a

##### **Initial prompt to understand how to use operators in Python (7)**

```python
# Step 1: Basic arithmetic operators with numbers
a = 1998
b = 2003
print("Addition:", a + b)       # Adds two numbers
print("Multiplication:", a * b) # Multiplies two numbers

# Step 2: Operators with strings
text = "Hello, world! "
print("String addition:", text + "It's 4 in the morning")  # Combines two strings
print("String multiplication:", text * 3)         # Repeats the string 3 times
```

**Step-By-Step**
a = 1998
b = 2003
print("Addition:", a + b)      
print("Multiplication:", a * b)
 - '+' adds two numbers (ex: 5 + 3 = 8).
 - '*' multiplies two numbers (ex: 5 × 3 = 15).
 - These are basic arithmetic operators.

text = "Hello, world! "
print("String addition:", text + "It's 4 in the morning")
print("String multiplication:", text * 3)
 - '+' with strings concatenates (joins) them
   - "Hello, world!" + "It's 4 in the morning" → "Hello, world! It's 4 in the morning".
 - '*' with a string and a number repeats the string
   - "Hello, world! " * 3 → "Hello, world! Hello, world! Hello, world! ".

**Key Aspects**

The same operator can behave differently depending on the data type
 - Numbers → '+' adds, '*' multiplies.
 - Strings → '+' joins, '*' repeats.
Python is type-sensitive. You cannot mix numbers and strings directly (e.g., "Hello" + 5 will cause an error). To combine text and numbers, you must convert numbers to strings using 'str()'. such as 'print("Number is: " + str(a))'
