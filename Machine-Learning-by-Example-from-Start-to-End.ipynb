## Critically Engaging with Social and Cultural Implications of AI 

AI technology has significantly improved work efficiency, but it also raises ethical issues such as data bias, transparency, and copyright. Based on a critical understanding of its social and cultural impacts, I will consider the risks and precautions associated with AI utilisation.

First, the most fundamental issue in utilising AI is data bias. Since AI outputs directly reflect the characteristics of its training and input data, biased data is likely to produce biased analysis results. When working with data biased toward specific cultures, languages, or backgrounds, there is a risk of mistakenly interpreting the results as general trends. As pointed out in ”Understand, Manage and Prevent Algorithmic Bias", this can lead to reinforcing social injustices and stereotypes. Therefore, when handling AI analysis results, it is essential to always be mindful of the data's context and limitations and make judgments cautiously.

Next is the issue of transparency and explainability. While this project also used AI tools like Copilot for code generation and task assistance, it is difficult for users to fully understand the rationale behind the suggestions made by such tools. The EU AI Act specifically requires disclosure when using AI, particularly for tasks involving generative AI, and demands explanations for the basis of outputs. In practice, this project ensured reproducibility and accuracy by having humans review and modify parts of the automatically generated code as needed. Thus, humans responsibly verifying AI outputs remains crucial for the safe utilisation of AI technology.

Furthermore, copyright and intellectual property issues cannot be ignored. Text or code generated by AI may resemble existing works, posing a risk of accidentally appropriating others' copyrighted material. When using generative AI, outputs should not be used directly. Instead, their content must be verified for issues and modified to align with the user's intent. Overreliance on AI could also diminish creativity and comprehension. Therefore, as the overall project lead, it is essential to critically evaluate AI suggestions rather than accepting them unquestioningly.

Considering these points, this project utilised AI technology while carefully considering its social and cultural impacts, aiming for ethically appropriate use. Future efforts to collect more diverse and unbiased data, increase transparency in AI usage, and enhance the explainability of results will enable more reliable analysis.

## Designing a Creative Mini Use of Large Language Models

#### What is a Large Language Model?

1. Training Data - LLMs are trained on large amounts of text from books, articles, websites, and other sources to learn patterns in human language.

2. Architecture - They use large neural networks with millions to billions of parameters, trained to predict the next word in a sequence.

3. Capabilities - They can perform a wide range of language tasks such as question answering, summarization, translation, and creative text generation.

4. Examples - Notable models include GPT-3/GPT-4 (OpenAI), BERT and T5 (Google), and the Llama series (Meta). GPT-4 powers Microsoft Copilot.

5. Applications - Used in chatbots, virtual assistants, content creation, customer service, and many other areas that involve understanding and generating human language.

The use of AI in cultural heritage and education facilitates access to information, but it also carries the risk of introducing bias into data and outputs. For example, models trained on data heavily influenced by specific cultural spheres or historical perspectives may reflect that bias in their results. Such bias not only risks giving users a false impression but also carries the risk of reinforcing social inequalities and stereotypes. Furthermore, perspectives on decolonisation, diversity, and inclusion are crucial. Cultural heritage projects using AI can enhance the representation and visibility of minority voices by incorporating the perspectives of underrepresented cultures and communities. This enables more equitable and inclusive information provision in educational and research settings. Additionally, transparency and explainability of AI are indispensable. Since outputs from Copilot or large language models (LLMs) may involve black-boxed internal processing and decision criteria, it is crucial to explicitly state the scope of AI used and the nature of any human modifications made. This ensures responsible use and enhances the reliability of results.

#### Comments on GPT-4, Llama, and Ollama

**Language Models can largely be used in three ways**

They can be used through APIs or coding libraries like GPT-4  which require coding skills. They can be included in applications with extra features, such as ChatGPT or Microsoft Copilot, that are available via cloud accounts. They can also be run as standalone executables like Llamafile on a local computer. Another option is to use platforms like HuggingFace, where pre-set models are available for people to try out and may require an account.



**In terms of data sharing concerns with respect to platforms**

Data-sharing with AI platforms raises concerns because services like ChatGPT are not transparent about how uploaded data is used, and simply making models larger with more data is not always better. Smaller models, such as TinyLlama, Llamafile, and Ollama, offer a way to work effectively with data locally on your own computer.
