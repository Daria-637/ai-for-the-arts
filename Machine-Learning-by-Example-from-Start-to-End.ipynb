## Reflecting on Machine Learning

Getting the code to run proved far more challenging than I had anticipated. This was largely because I was practically new to using Jupyter and GitHub, and I stumbled right from the basics – figuring out where everything was located within the environment and how to operate it.

Even during the initial setup phase, I struggled to keep up with the concepts. I repeatedly watched YouTube tutorial videos, read “Title Learn Python 3 the Hard Way: A Very Simple Introduction to the Terrifyingly Beautiful World of Computers and Code” by Shaw, used Copilot and Google to research each step individually, made mistakes, corrected them, tried again, and then fixed them once more, going through this process countless times.

While writing the code itself, I frequently got stuck because I could not understand why errors were occurring. Particularly, the mechanism requiring ”pip install" when a library was needed was something I only grasped after considerable trial and error. Furthermore, when adding files from a repository, I had no idea about the basic yet crucial rule that you must open the Codespace from the Main branch to perform a “Commit & Sync”. Unable to find the reason for the failure, I genuinely worried for about an hour that I might have broken the system, finding myself utterly at a loss.

What surprised me even more was that simply adding code resulted in images being generated in ways I had never imagined. Witnessing unintended behaviour occur, I experienced firsthand how precisely code responds to instructions and processes data. It made me feel both the complexity and the fascination of the programming world simultaneously.

Thus, the process was far from smooth from start to finish. Rather, I frequently felt anxiety and frustration whenever encountering obstacles. However, through these experiences, I gradually grasped GitHub's fundamental operations and the underlying principles of how code functions. Though I felt my resolve waver many times, persisting without giving up taught me the tenacity to see things through to the end and cultivated a mindset for tackling problems head-on. Looking back now, this process itself proved an invaluable learning experience.
